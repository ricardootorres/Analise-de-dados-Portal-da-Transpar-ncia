```python
import pandas as pd
```

```python
df_viagens = pd.read_csv('../data/raw/2023_Viagem.csv', encoding='Windows-1252', sep=';', decimal=',', thousands='.')
```

```python
df_viagens['Despesas'] = df_viagens['Valor diárias'] + df_viagens['Valor passagens'] - df_viagens['Valor devolução'] + df_viagens['Valor outros gastos']
df_viagens[['Período - Data de início', 'Período - Data de fim']] = df_viagens[['Período - Data de início', 'Período - Data de fim']].apply(pd.to_datetime, format='%d/%m/%Y')
df_viagens['Duracao'] = (df_viagens['Período - Data de fim'] - df_viagens['Período - Data de início']).dt.days
```

### Pergunta 1: Órgãos que mais gastam

```python
df_viagens.groupby('Nome do órgão superior')['Despesas'].sum().sort_values(ascending=False).head(10)
```

### Pergunta 6: Cargos que mais viajam

```python
df_viagens['Cargo'].value_counts().head()
```

### Pergunta 8: Cargos com mais despesas

```python
df_viagens.groupby('Cargo')['Despesas'].sum().sort_values(ascending=False).head(10)
```

### Pergunta 7: Duração média

```python
df_viagens['Duracao'].mean()
```
